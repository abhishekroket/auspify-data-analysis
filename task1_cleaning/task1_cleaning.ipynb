```python

import pandas as pd
df = pd.read_csv("../data/netflix_titles.csv")
print(df.shape)
df.info()
df.head()
```

```python
print(df.isnull().sum())


placeholders = ["not given", "unknown", "n/a", "nan", ""]
for col in ["director", "country", "rating"]:
    count = df[col].str.strip().str.lower().isin(placeholders).sum()
    print(f"{col}: {count} placeholder values")

for col in ["director", "country"]:
    mask = df[col].str.strip().str.lower().isin(placeholders)
    df.loc[mask, col] = "Unknown"
```

```python
print("Full-row duplicates:", df.duplicated().sum())
print("Duplicate show_id:", df.duplicated(subset="show_id").sum())
df = df.drop_duplicates()
df = df.drop_duplicates(subset="show_id")

for col in ["type", "title", "director", "country", "rating", "listed_in", "duration"]:
    df[col] = df[col].str.strip()

df["date_added"] = pd.to_datetime(df["date_added"], format="%m/%d/%Y", errors="coerce")
print("Unparsed dates:", df["date_added"].isnull().sum())
df["year_added"] = df["date_added"].dt.year
df["month_added"] = df["date_added"].dt.month_name()

df[["duration_value", "duration_unit"]] = df["duration"].str.extract(r"(\d+)\s*(\w+)")
df["duration_value"] = df["duration_value"].astype(int)
```

```python
df["type"] = df["type"].str.title().replace({"Tv Show": "TV Show"})
df["country"] = df["country"].str.title()
df["rating"] = df["rating"].str.upper()

print(df["type"].value_counts())
print(df["rating"].value_counts())
print("Countries containing a comma:", df["country"].str.contains(",").sum())
print(df["country"].nunique(), "unique countries")
```

```python
print(df.shape)
df.to_csv("../data/netflix_cleaned.csv", index=False)
print(df.isnull().sum().sum(), "missing values left")
```
